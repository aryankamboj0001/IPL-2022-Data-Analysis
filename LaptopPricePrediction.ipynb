```python
df['Ram'] = df['Ram'].str.replace('GB','').astype(int)
df['Weight']= df['Weight'].str.replace('kg','').astype(float)
```

```python
df.head()
```

```python
for col in df.columns:
  print(f"Value counts for column: {col}")
  print(df[col].value_counts())
  print("\n")
```

```python
df.columns
```

```python
category = ['Company','ScreenResolution','TypeName','Memory','Gpu','OpSys','Cpu']
```

```python
df_encoded = pd.get_dummies(df, columns =category,drop_first=True,dtype=int)
```

```python
df_encoded.drop(columns=['laptop_ID','Product'],inplace=True)
```

```python
df_encoded
```

```python
df_encoded.info()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
numerical_features = ['Inches','Ram','Weight']
df_encoded[numerical_features] = scaler.fit_transform(df_encoded[numerical_features])
```

```python
exchange_rate_eur_to_inr = 90  # Approximate exchange rate
df_encoded['Price_inr'] = df_encoded['Price_euros'] * exchange_rate_eur_to_inr
df_encoded.head()
```

```python
df_encoded.drop(columns=['Price_euros'],inplace=True)
```

```python
from sklearn.model_selection import train_test_split
X = df_encoded.drop(columns=['Price_inr'])
y = df_encoded['Price_inr']
```

```python
from sklearn.preprocessing import StandardScaler

y_scaler = StandardScaler()

y = y_scaler.fit_transform(df_encoded['Price_inr'].values.reshape(-1,1))
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
#Building and ANN
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
input_dim = X_train.shape[1]
```

```python
input_dim
```

```python
model = Sequential()
model.add(Dense(units = 128,activation='relu',input_shape=(input_dim,)))
model.add(Dropout(0.2))    #1 hidden layers
model.add(Dense(units = 128, activation = 'relu'))
model.add(Dropout(0.2))    #2 hidden layers
model.add(Dense(units = 64,activation = 'relu') )
model.add(Dropout(0.2))          #3 hidden layers
model.add(Dense(units = 32,activation = 'relu') )
model.add(Dropout(0.2))          #4 hidden layers
```

```python
model.add(Dense(units = 1 , activation = 'linear'))
```

```python
model.summary()
```

```python
early = EarlyStopping(monitor='val_loss',patience=5,verbose=1)
```

```python
model.compile(optimizer='adam',loss='mean_squared_error',metrics=['mean_absolute_error'])
```

```python
history = model.fit(X_train,y_train,epochs=100,batch_size = 32,validation_split=0.2,verbose=1,callbacks=[early])
```

```python
y_pred = model.predict(X_test)

y_pred_real = y_scaler.inverse_transform(y_pred)
```

```python
y_pred_real
```

```python
y_pred_trained = model.predict(X_train)
y_pred_trained_real = y_scaler.inverse_transform(y_pred_trained)
```

```python
y_pred_trained_real
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_test_real = y_scaler.inverse_transform(y_test)

mse = mean_squared_error(y_test_real, y_pred_real)
mae = mean_absolute_error(y_test_real, y_pred_real)
r2 = r2_score(y_test_real, y_pred_real)

print("MAE:", mae)
print("MSE:", mse)
print("R2:", r2)
```

```python
plt.plot(history.history['loss'],label='train loss')
plt.plot(history.history['val_loss'],label='val loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
```
